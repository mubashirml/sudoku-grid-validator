# 🧩 Sudoku Grid Validator using NumPy

A practical NumPy project for validating a 9×9 Sudoku grid using:

- NumPy arrays
- Array indexing and slicing
- Row and column validation
- 3×3 block extraction
- Array sorting and comparison
- Boolean validation logic

### Objective

To verify whether a completed Sudoku grid satisfies all
standard row, column, and 3×3 block constraints.

```python
# ============================================================
# 01. IMPORT NUMPY
# ============================================================

import numpy as np

print("=" * 55)
print("              SUDOKU GRID VALIDATOR")
print("=" * 55)
print("NumPy imported successfully.")
```

```python
# ============================================================
# 02. CREATE SUDOKU GRID
# ============================================================

sudoku = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],

    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],

    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])

print("Sudoku grid created successfully.")
print(f"Grid Shape : {sudoku.shape}")
print(f"Total Cells: {sudoku.size}")
```

```python
# ============================================================
# 03. DISPLAY SUDOKU GRID
# ============================================================

print("\n" + "=" * 55)
print("                    SUDOKU GRID")
print("=" * 55)

for row_index, row in enumerate(sudoku):

    if row_index % 3 == 0 and row_index != 0:
        print("-" * 21)

    print(
        f"{row[0]} {row[1]} {row[2]} | "
        f"{row[3]} {row[4]} {row[5]} | "
        f"{row[6]} {row[7]} {row[8]}"
    )

print("=" * 55)
```

```python
# ============================================================
# 04. CALCULATE OVERALL SUM
# ============================================================

overall_sum = np.sum(sudoku)

print("\n" + "=" * 55)
print("                GRID SUMMARY")
print("=" * 55)

print(f"Total number of cells : {sudoku.size}")
print(f"Overall grid sum      : {overall_sum}")
print("=" * 55)
```

```python
# ============================================================
# 05. VALIDATE ROWS
# ============================================================

expected_numbers = np.arange(1, 10)

valid_rows = True

print("\n" + "=" * 55)
print("                  ROW VALIDATION")
print("=" * 55)

for index, row in enumerate(sudoku, start=1):

    sorted_row = np.sort(row)

    if np.array_equal(sorted_row, expected_numbers):
        print(f"✅ Row {index:02d}: Valid")
    else:
        print(f"❌ Row {index:02d}: Invalid")
        valid_rows = False

print("-" * 55)

if valid_rows:
    print("RESULT: All 9 rows are valid.")
else:
    print("RESULT: One or more rows are invalid.")

print("=" * 55)
```

```python
# ============================================================
# 06. VALIDATE COLUMNS
# ============================================================

valid_columns = True

print("\n" + "=" * 55)
print("                COLUMN VALIDATION")
print("=" * 55)

for index, column in enumerate(sudoku.T, start=1):

    sorted_column = np.sort(column)

    if np.array_equal(sorted_column, expected_numbers):
        print(f"✅ Column {index:02d}: Valid")
    else:
        print(f"❌ Column {index:02d}: Invalid")
        valid_columns = False

print("-" * 55)

if valid_columns:
    print("RESULT: All 9 columns are valid.")
else:
    print("RESULT: One or more columns are invalid.")

print("=" * 55)
```

```python
# ============================================================
# 07. EXTRACT 3x3 BLOCKS
# ============================================================

blocks = []

for row in range(0, 9, 3):
    for column in range(0, 9, 3):

        block = sudoku[row:row + 3, column:column + 3]
        blocks.append(block)

print("\n" + "=" * 55)
print("              SUDOKU 3x3 BLOCKS")
print("=" * 55)

for index, block in enumerate(blocks, start=1):

    print(f"\nBlock {index:02d}:")
    print(block)

print("=" * 55)
```

```python
# ============================================================
# 08. VALIDATE 3x3 BLOCKS
# ============================================================

valid_blocks = True

print("\n" + "=" * 55)
print("                BLOCK VALIDATION")
print("=" * 55)

for index, block in enumerate(blocks, start=1):

    flattened_block = block.flatten()
    sorted_block = np.sort(flattened_block)

    if np.array_equal(sorted_block, expected_numbers):
        print(f"✅ Block {index:02d}: Valid")
    else:
        print(f"❌ Block {index:02d}: Invalid")
        valid_blocks = False

print("-" * 55)

if valid_blocks:
    print("RESULT: All 9 blocks are valid.")
else:
    print("RESULT: One or more blocks are invalid.")

print("=" * 55)
```

```python
# ============================================================
# 09. COMPLETE SUDOKU VALIDATION
# ============================================================

sudoku_is_valid = (
    valid_rows
    and valid_columns
    and valid_blocks
)

print("\n" + "=" * 55)
print("            COMPLETE VALIDATION")
print("=" * 55)

print(f"Rows       : {'VALID' if valid_rows else 'INVALID'}")
print(f"Columns    : {'VALID' if valid_columns else 'INVALID'}")
print(f"3x3 Blocks : {'VALID' if valid_blocks else 'INVALID'}")

print("-" * 55)

if sudoku_is_valid:
    print("STATUS: Sudoku grid is completely valid. ✅")
else:
    print("STATUS: Sudoku grid is invalid. ❌")

print("=" * 55)
```
